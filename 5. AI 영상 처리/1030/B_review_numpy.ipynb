# Numpy 라이브러리 -> 데이터 연산 또는 관리에 사용

```python
# np.array() :: numpy 배열 생성하는 기초 함수
num = [15, 25, 35, 45]
nnum = np.array([15, 25, 35, 45])

print(num * 5)
print(nnum * 5)   # 4개의 요소에 대해 산술
```

```python
data = np.array([
    [25, 50, 75],
    [125, 150, 175]
])  # 2행 3열의 numpy 배열 생성
print(data / 5)
```

```python
arr = np.array(
    [range(i, i+10, 2) for i in [1, 11, 111]],
    dtype = np.int32
)
# i=1   -> 1, 3, 5, 7, 9
# i=11  -> 11, 13, 15, 17, 19
# ...
print(arr)
```
