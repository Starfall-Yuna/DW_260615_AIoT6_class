```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. 배열 생성 함수 :: array(), zeros(), ones(), full(), arange(), eye()
# 2. 통계 함수 :: mean(), max(), min(), sum()
# 3. 조건에 따른 배열 처리 :: where()
# 4. 배열 범위 제한 :: clip()
# 5. 배열값 연산 :: 브로드캐스팅
# 6. 배열 형태, 자료형 변형 :: reshape(), astype()
# 7. 배열값 -> 이미지 시각화
```

```python
image = np.array([
    [ 20,  40,  60,  80, 100, 120],
    [ 40,  60,  80, 100, 120, 140],
    [ 60,  80, 100, 120, 140, 160],
    [ 80, 100, 120, 140, 160, 180],
    [100, 120, 140, 160, 180, 200],
    [120, 140, 160, 180, 200, 220]
])

# 1. 전체 평균 픽셀값 구하기    -> mean()
avg = np.mean(image)
print('평균:', avg, '\n')

# 2. 전체 평균 미만인 픽셀값에는 50을 더하고, 이상인 픽셀값에는 30을 빼기 -> where()
result = np.where(image < avg, image + 50, image - 30)
print(result, '\n')

# 3. 픽셀값 범위 제한 (0 ~ 255)    -> clip()
result = np.clip(result, 0, 255)
print(result, '\n')

# 4. 가운데 2*2 영역을 흰색(255)으로 변경
result[2:4, 2:4] = 255  # 2행2열 ~ 4행4열 이전까지
print(result, '\n')

# 5. 최종 배열의 최댓값, 최솟값을 출력
print('최댓값:', np.max(result))
print('최솟값:', np.min(result))

# 6. 원본 이미지와 보정된 이미지를 각각 흑백 시각화 (이미지 보여주기)
plt.figure(figsize = (10, 5))

  # 원본 표현
plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray', vmin=0, vmax=255)
plt.title('Original')

  # 보정 이미지 표현
plt.subplot(1, 2, 2)
plt.imshow(result, cmap='gray', vmin=0, vmax=255)
plt.title('Modify')
```

```python
# 보충
print(np.zeros([2, 4]), '\n')
print(np.ones([2, 4]), '\n')
print(np.full([2, 4], 3), '\n')
print(np.eye(3), '\n')
```

```python
# 보충
arr = np.array([
    [1, 5, 9],
    [10, 50, 90]
])

print(np.sum(arr))
print(np.sum(arr, axis=0))  # 세로(열) 기준으로의 합계
print(np.sum(arr, axis=1))  # 가로(행) 기준으로의 합계
```

```python
# 보충
print(arr)
print(arr.reshape(3, 2))  # 2행3열 -> 3행2열
print(arr.astype(np.float32))
```
