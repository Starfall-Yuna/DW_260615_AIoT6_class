

*   각 블럭마다 실행 독립적
*   메모리(변수, 함수, 라이브러리, ...)는 한 페이지 내에서 공유됨



```python
class Video:
  # 생성자
  def __init__(self):
    # 필드(멤버변수) 2개 생성, 값 설정
    self.minute = 60
    self.file = '.mp4'
```

```python
v = Video()
print(v.minute, '분짜리', v.file, '형식 파일')
```

```python
# 예외처리 (try - except - finally)

try:
  n1 = int(input('첫번째 정수 입력 > '))
  n2 = int(input('두번째 정수 입력 > '))
  print(n1 / n2)
except ValueError:
  # 사용자가 정수형 입력을 하지 않았을 때, 여기로 이동
  # '3.555'(문자열) -> 바로 정수로 변환X
  print('정수만 입력해주세요.')
except ZeroDivisionError:
  # n2가 0으로 입력받을 때, 여기로 이동
  print('0으로 나눌 수 없습니다')
finally:
  # 에러 발생 여부 상관없이, 마지막에 실행
  print('프로그램을 종료합니다')
```
