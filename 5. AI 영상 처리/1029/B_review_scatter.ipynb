```python
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
```

# Matplotlib 라이브러리 활용
*   데이터 시각화, 그래프/차트 표현에 사용

# Tips 데이터 표 -> 시각화 진행


*   scatter() 산점도 그래프 표현
*   데이터를 점으로 표현했을 때 <br>
    점의 밀집도를 통해서 2개 속성 간의 관계 파악



```python
# 데이터 표 불러오기
tips = sns.load_dataset('tips')

# 데이터 확인
print(tips.head(10))    # tips의 상위 10개 데이터 불러오기
print(tips.shape)       #       244개의 행(데이터), 7가지의 열(속성)
print(tips.columns)     #       속성 나열
```

```python
# "팁의 액수(tip)가 크면, 총 결제 금액(total_bill)도 커질 수 있는가?"

plt.figure(figsize=(7, 5))
plt.scatter(tips['tip'], tips['total_bill'], alpha=0.6, color='green')
plt.xlabel('tip')
plt.ylabel('total_bill')
plt.title('tip -> total_bill')
plt.grid(linestyle = '--')
plt.show()

# 팁의 액수가 클수록 (오른쪽으로 갈수록)
# 총 결제 금액도 커지는 경향이 보임 (점들이 위로 올라간다)
#   => 관계O / 질문의 해답 : 참
```

```python
# "흡연자(smoker)일수록 팁(tip)을 더 많이 주는 경향이 있는가?"

# 그래프 그리기
plt.figure(figsize=(7, 5))    # 평면도 표현할 캔버스 생성
plt.scatter(tips['smoker'], tips['tip'], alpha = 0.6, color='green')
  # x축(가로축)을 흡연자 여부 설정, y축(세로축)을 팁 액수 설정
  # 점의 불투명도 60%, 점의 색상을 초록색
plt.xlabel('smoker')    # x축, y축 텍스트 설정
plt.ylabel('tip')
plt.title('smoker -> tip')  # 그래프 이름 붙이기
plt.grid(linestyle='--')    # 평면도 눈금자 표시
plt.show()

# 관계성 파악
#   흡연자인 경우 - 흡연자가 아닌 경우 그래프 형태 비슷함
#     => 관계성X
```
