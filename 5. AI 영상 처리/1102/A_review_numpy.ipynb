# Numpy 라이브러리 :: 연산과 산술에 활용되는 배열 관리

```python
# 2차원 numpy 배열 생성
arr = np.array([
  [10, 20, 30],
  [40, 50, 60]
])
print(arr)
```

```python
print(arr.dtype)    # 데이터 타입 확인
print(arr.shape)    # 2행 3열인 배열의 모양 확인
print(arr.size)     # 6개의 배열값 개수 확인
print(arr.ndim)     # 2차원 배열에 대해 확인 ([]쌍이 몇 개 중첩되었는지)
```

```python
# [10, 20, 30],
# [40, 50, 60]

print(arr[0, :])    # 0행(가로)의 전체(:) 값 슬라이싱
print(arr[:, 1])    # 1열(세로)의 전체 값 슬라이싱
print(arr[0:2, :])  # 0행~2행 이전까지의 전체 값 슬라이싱
print(arr[0:2, 0:2])  # 0행~2행 이전까지, 0열~2열 이전까지 (2열(30, 60) 포함X)
```

```python
arr[0, :] = 1   # [10, 20, 30] --변경--> [1, 1, 1]
print(arr)
```

```python
# 5행 5열로 요소의 값을 모두 0으로 설정한 numpy 배열을 생성하고   -> zeros() 활용
# 아래와 같은 형태로 값을 변경하여라 (반복문 사용X, 슬라이싱만 사용)
# [[0 0 8 0 0]
#  [0 0 8 0 0]
#  [8 8 8 8 8]
#  [0 0 8 0 0]
#  [0 0 8 0 0]]

arr = np.zeros((5, 5), dtype=int)

arr[2, :] = 8     # 2행의 모든 값, 8로 설정
arr[:, 2] = 8     # 2열의 모든 값, 8로 설정

print(arr)
```
