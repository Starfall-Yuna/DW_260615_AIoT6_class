```python
import numpy as np
import matplotlib.pyplot as plt
```

# 배열 생성 함수

```python
np.zeros([2, 3])    # 2행 3열, 모두 0값인 배열 생성
```

```python
np.zeros([2, 3]) + 5    # 모든 위치에 +5 수행
```

```python
np.ones([2, 3])     # 2행 3열, 모두 1값인 배열 생성
```

```python
np.ones([2, 3]) * 255   # 모든 위치에 *255 수행
```

```python
np.full([2, 3], -1)     # 2행 3열, 모두 -1값인 배열 생성
```

```python
np.arange(0, 10, 2)     # 0부터 10 이전까지, +2씩 간격을 낸 값들로 배열 생성
```

```python
np.eye(5)         # 5행 5열, 주 대각선을 이루는 위치는 모두 1로, 나머지는 0으로
  # (0, 0), (1, 1), (2, 2), ...   => 이 값들을 모두 1로
```

# 통계 함수

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
print(arr)

print(np.mean(arr))   # 평균 확인
print(np.sum(arr))    # 합계 확인
print(np.min(arr))    # 최솟값 확인
print(np.max(arr))    # 최댓값 확인
```

```python
# [10, 20, 30],
# [40, 50, 60],
# [70, 80, 90]
# arr에 대해서

# 열마다 평균 계산하는 코드 한줄
print(np.mean(arr, axis = 0))
# 행마다 평균 계산하는 코드 한줄
print(np.mean(arr, axis = 1))
```

# 조건에 따른 배열 처리

```python
# [10, 20, 30],
# [40, 50, 60],
# [70, 80, 90]

print(arr > 30)   # 각 위치의 값마다, 30 초과인지에 따라 참/거짓 판별
print(arr[arr > 30])    # 조건식 만족 요소들만 불러옴

arr[arr > 30] = 100     # 조건식 만족 요소들에 대해, 값 설정
print(arr)
```

```python
image = np.array([
    [25, 75, 125, 175],
    [45, 95, 145, 195],
    [65, 115, 165, 215],
    [85, 135, 185, 235]
])

# 주어진 배열에 있어, 아래와 같은 연산을 수행하라.
# 1. 150 이상인 값 출력
print(image[image >= 150])
  # 읽는 순서 :: 0행(0열 -> 1열, ...) -> 1행 -> 2행 -> ...

# 2. 100 미만을 0으로 변경
image[image < 100] = 0

# 3. 200 이상을 255로 변경
image[image >= 200] = 255

# 4. 최종 배열 출력
print(image)
```

```python
x = np.array([10, 20, 30, 40])
print(x)

# np.where(condition, a, b) :: (삼항연산자 유사)
#   조건식에 만족하면 a로 변환, 아니면 b로 변환
result = np.where(x > 25, 100, 0)
print(result)

r2 = np.where(x > 25, 100, x)   # x>25 만족하면 100 변경, 아니면 기존 값 유지
print(r2)
```

# 배열값 연산 (범위 제한)

```python
arr = np.array([-1, 100, 260, 300])
r = np.clip(arr, 0, 255)      # 배열값 범위를 0 ~ 255 제한 (범위를 벗어난 값은 영역의 최솟값/최댓값 변경)
print(r)
```

```python
# 문제
image = np.array([
    [40, 80, 120],
    [100, 160, 220],
    [150, 210, 250]
])

# 1. 모든 값에 70 더하기
image += 70

# 2. 값의 범위를 0~255로 제한
result = np.clip(image, 0, 255)

# 3. 결과 출력
print(result)
```

```python
# 브로드캐스팅 :: "배열의 차원/형태가 달라도, (규칙이 보이면?) 산술적인 연산을 수행O"
base = np.array([
    [10, 20, 30],
    [100, 200, 300]
])
calc1 = np.array([3])
print(base + calc1)   # 2차원 + 1차원 => 정상 실행

calc2 = np.array([3, 6])
# print(base + calc2)     # [10, 20, 30] + [3, 6]   => 규칙적인 덧셈X -> 에러 발생

calc3 = np.array([3, 6, 9])
print(base + calc3)   #30] + [3, 6, 9]  => 규칙적인 덧셈O [10, 20,
```

# 배열 자료형/형태 변환

```python
arr = np.array([1, 2, 10, 20, 100, 200])
print(arr)

arr_2d = arr.reshape(3, 2)    # 1차원 배열 -> 3행 2열의 2차원 배열 변환
print(arr_2d)

# arr_2d를 2행 3열로 변환
print(arr.reshape(2, 3))
```

```python
arr = np.array([1.2, 2.5, 3.8])   # float형 numpy 배열
print(arr.astype(np.int32))     # 32비트 int형 자료형 변환
```

# Numpy 배열값 기반 이미지 시각화

```python
img = np.zeros([5, 5])

# 2행의 전체 값(중앙 가로선)을 255로 설정
#img[2, :] = 255

# 중앙에 있는 9칸에 대해서 노란 색상 변경
img[1:4, 1:4] = 255

# plt.imshow(img, cmap = 'gray', vmin = 0, vmax = 255)
  # 흑백 톤의 화면으로 표시
  # 값의 범위 :: 0 ~ 255
# plt.imshow(img, cmap = 'jet', vmin = 0, vmax = 255)
plt.imshow(img, cmap = 'viridis', vmin = 0, vmax = 255)
  # cmap 기본값 :: 'viridis'
plt.show()
```
