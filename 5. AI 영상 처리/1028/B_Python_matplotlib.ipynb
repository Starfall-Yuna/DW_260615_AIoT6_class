# * Matplotlib (데이터 시각화)
그래프, 차트, ... 그리는 라이브러리

```python
# 필요한 라이브러리 적용 (자동 다운로드)
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
```

```python
# y = 3*x + 5 그래프 그리기
# y = x^2 - 1 그래프 그리기

# 1. 수식을 함수(f, g)로 구성
def f(x):
  return 3*x + 5
def g(x):
  return x**2 - 1

# 선그래프 표현 <- "여러 점을 구성"

# 2. 여러 개의 점의 x값 생성
#   (-5부터 5.1 이전까지, 0.5씩 증가하며 x값 생성)
x = np.arange(-5, 5.1, 0.5)
print(x)

# 3. 2.에 대응하는 y값 생성
y = f(x)    # 3*x + 5에 대한 결과값
print(y)
gy = g(x)   # x^2 -1에 대한 결과값
print(gy)

# 4. 최종적으로, 그래프 그리기
plt.plot(x, y)    # f(x) 그래프 그리기
plt.plot(x, gy)   # g(x) 그래프 그리기
plt.xlim(-6, 6)     # x축 영역 범위
plt.ylim(-15, 30)   # y축 영역 범위
plt.grid(linestyle='--')  # 평면도 격자 표시
plt.show()
```

```python
# 평면도에 원 그리기

# 원을 그릴 때 필요한 정보 :: 원의 중심(x좌표, y좌표), 반지름
x = 0
y = 0
r = 3

# patches를 통해, 원 그리기 수행
print('제공된 정보로 원을 그립니다')
c = patches.Circle((x, y), r, fill = False, color = 'b')
plt.gca().add_patch(c)    # c의 정보를 기반으로 좌표에 그리기

# 평면도 출력
plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.show()
```

```python
# seaborn 라이브러리 :: 다량의 통계 데이터 활용
tips = sns.load_dataset('tips')
tips.head()

# tips 데이터 표의 속성들
# 결제총액 / 팁 액수 / 성별 / 흡연자여부 / 요일 / 시간대(점심/저녁) / 인원수
```

```python
print(tips.head(10))    # 상위 10개 가져옴
print(tips.shape)       # 244개의 데이터, 7개의 속성
print(tips.columns)     # 속성들을 하나씩 출력
```

# 그래프 그리기 :: scatter() 산점도 그래프 그리기
(그래프 표현하는 이유 :: 2가지 데이터 상관관계 파악)

```python
# "결제 금액(total_bill)이 많을수록 팁(tip)도 함께 많아지는가?"

# tips 표 안의 244개 데이터를 전부 점으로 표현
plt.figure(figsize=(7, 5))      # 평면도 캔버스 설정
plt.scatter(tips['total_bill'], tips['tip'], alpha=0.6, color = 'coral')
    # 결제총액(total_bill)을 가로축, 팁금액(tip)을 세로축으로 기준잡아 그래프 그림
    # 불투명도를 60%로 설정, 점의 색상을 코랄로 지정
plt.xlabel('Total Bill')    # 가로축, 세로축 설명 지정
plt.ylabel('Tip')
plt.title('Total Bill - Tip')   # 그래프 제목 지정
plt.grid(linestyle='--')        # 평면도 눈금 표시
plt.show()

# 결제 금액이 많아질수록, 팁 금액 또한 함께 올라가는 형상 발견
#   결제 금액이 커질수록, 팁 금액도 함께 커진다고 볼 수 있음
```

```python
# [문제1] "식사 인원수(size)가 많을수록 총 결제 금액(total_bill)도 함께 증가하는가?"

# 그래프 출력
plt.figure(figsize=(7, 5))
plt.scatter(tips['size'], tips['total_bill'], alpha=0.6)
plt.xlabel('size')
plt.ylabel('total_bill')
plt.title('size -> total_bill')
plt.grid(linestyle='--')
plt.show()

# 문제에 대한 답변을 그래프 기반으로 진행
#     => 식사 인원수가 많을수록 총 결제 금액이 대게 증가
#     => 상관관계 존재O
```

```python
# [문제2] "식사 인원수(size)가 많을수록 팁 금액(tip)도 함께 증가하는가?"

# 그래프 출력
plt.figure(figsize=(7, 5))
plt.scatter(tips['size'], tips['tip'], alpha=0.6)
plt.xlabel('size')
plt.ylabel('tip')
plt.title('size -> tip')
plt.grid(linestyle='--')
plt.show()

# 문제에 대한 답변을 그래프 기반으로 진행
#   => 식사 인원수가 증가할수록 팁 금액도 대개 함께 증가
#   => 상관관계 존재O
```
