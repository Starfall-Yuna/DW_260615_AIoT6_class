```python
# 아래의 코드 작동에 영향을 줌
score = 55
```

```python
# 위의 score 변수 생성부터 "실행"하기
if score >= 90:
  print('A')
elif score >= 80:
  print('B')
elif score >= 70:
  print('C')
elif score >= 60:
  print('D')
else:
  print('F')
```

```python
def odd_even(num):
  if num%2 == 0:
    return '짝수'
  else:
    return '홀수'
print(odd_even(50))
print(odd_even(49))
```
