# 머신러닝 (Machine-Learning, 기계 학습)
(딥러닝이 머신러닝의 일종)
*   과거 데이터에서 발견될 만한 규칙/패턴 기반
*   기존 데이터 이외에 대해 결과값 예측

```python
import numpy as np
import matplotlib.pyplot as plt

# scikit-learn (여러 머신러닝 모델 정보 보유)
# "선형회귀 모델"
from sklearn.linear_model import LinearRegression
```

```python
# 과거 데이터 준비
compress = np.array([0, 20, 40, 60, 80])        # 파일 압축 비율(%) (예시 입력 데이터)
size = np.array([310, 250, 210, 160, 110])      # 이미지 파일 크기(MB) (예시 결과 데이터)

# 파일 압축 비율 - 이미지 파일 크기 => 반비례 관계
plt.scatter(compress, size, color='blue')
plt.xlabel('Compress')
plt.ylabel('size')
plt.show()

# 두 데이터가 반비례 관계임을 파악할 수 있도록, 머신러닝 학습 진행
```

```python
x = compress.reshape(-1, 1)

# Expected 2D array, got 1D array instead   -> 요인의 배열은 2차원으로

# LinearRegression() :: "비어있는" 선형회귀 모델 생성
model = LinearRegression()    # 학습할 수 있는 환경 만들기
# fit() :: 데이터를 삽입하여 학습 진행
model.fit(x, size)            # 실제 학습
```

```python
# Expected 2D array, got 1D array instead

# predict() :: 특정 요인에 대한 예측값 제시
# 압축 비율이 30%일 때, 이미지 파일 크기 예측값 받아옴
result = model.predict([[30]])
print('예측 결과:', result)
```

```python
# 머신러닝 예측 결과 그래프 그리기
test_input = np.arange(0, 101, 10)      # [0, 10, 20, ..., 90, 100]
pred = model.predict(test_input.reshape(-1, 1))
  # [0, 10, 20, ..., 90, 100]에 대한 11개의 예측값 불러오기

plt.scatter(compress, size, color='blue', label='Original')
plt.plot(test_input, pred, color='red', label='Predict')

plt.xlabel('Compress(%)')
plt.ylabel('Size(MB)')
plt.legend()
plt.show()

# 파란색 점-빨간색 점 간극 => "오차값"
```
