# 인공지능 -> 머신러닝(Machine-learning, 기계 학습)

*   기존 데이터 기반으로 "규칙, 패턴" 찾아내서
*   이를 기반으로 미래의 새로운 데이터 예측


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 문제 :: "공부 시간(x)이 길수록 시험 점수(y)가 높아지는가?"

# * 기존 과거 데이터 준비
#     [가정] 1시간 공부하면 50점, 2시간 공부하면 60점, ....
x = np.array([1, 2, 3, 4, 5])
y = np.array([50, 60, 65, 75, 85])
print('공부시간:', x)
print('시험점수:', y)

# * 그래프 그리기
plt.scatter(x, y)
plt.xlabel('Study Time')
plt.ylabel('Score')
plt.show()
```

```python
# 선형 회귀 모델
#   => "데이터의 상관관계" 기반으로, 미래 데이터 예측
from sklearn.linear_model import LinearRegression
```

```python
# 머신러닝 활용하기 전, 사전 작업

# x = np.array([1, 2, 3, 4, 5])   -> 1행 5열
x = x.reshape(-1, 1)            # -> 5행 1열 (2차원 배열 변형)
print(x)

# "공부시간과 수면시간이 모두 시험 점수에 영향을 미치는가?"
# 1시간 공부하고 7시간 잤음 -> 50점
# 2시간 공부하고 7시간 잤음 -> 60점, ...
# x = [
#     [1, 7],   -> y = 50
#     [2, 7],   -> y = 60
#     [3, 6],   -> y = 65
# ]
# 요인이 2개 이상인 경우 -> 2차원 배열 구성해야함

# 결과에 영향을 주는 요인의 데이터 구분지어야 함
```

```python
# 머신러닝 모델 생성
model = LinearRegression()
# 모델 학습
model.fit(x, y)     # x(공부시간(요인)), y(시험점수(결과))
```

```python
# 예측값을 result에 저장
result = model.predict([[6]])

print('7시간 공부하면 예측 점수:', result)
```

```python
# 예측 그래프에 표현할 데이터 불러오기
test = np.arange(1, 8)      # test = [1, 2, ..., 7]
result = model.predict(test.reshape(-1, 1))

plt.scatter(x.ravel(), y, color='blue')     # 점 형태, 기존 데이터
plt.plot(test, result, color='red')         # 선 형태, 모델의 예측값

plt.xlabel('Study Time')
plt.ylabel('Score')
plt.show()
```
