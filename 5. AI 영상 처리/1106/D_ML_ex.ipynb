```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# "배달 거리(x)가 길면, 배달 시간(y)도 길어지는가?"

# 1. 기존 데이터 준비
distance = np.array([1, 2, 3, 4, 5])     # 배달 거리(km)
time = np.array([6, 15, 21, 27, 35])   # 배달 시간(m)

print(distance)
print(time)
```

```python
# 2. x 데이터 형태 변형 (1행5열(1차원) -> 5행1열(2차원))
x = distance.reshape(-1, 1)
print(x)
```

```python
# 3. 모델 생성 + 데이터 삽입하여 학습
model = LinearRegression()
model.fit(x, time)    # 요인(x) -> 결과(time)
```

```python
# 기존 데이터 + 예측 데이터 시각화 -> 비교
plt.scatter(distance, time)

test_x = np.arange(1, 10)     # 1km ~ 9km 거리 데이터 삽입
pre_result = model.predict(test_x.reshape(-1, 1))   # 1~9km 거리에 대한 배달 시간 예측값
# plt.scatter(test_x, pre_result)
plt.plot(test_x, pre_result, color='red')

plt.xlabel('dis')
plt.ylabel('time')
plt.show()
```
