```python
import matplotlib.pyplot as plt

from sklearn.datasets import load_digits
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
```

```python
# * 데이터 불러오기

# 손글씨로 숫자를 적은 이미지 불러오기 (1797개)
digits = load_digits()
print(digits)
```

```python
# * 머신러닝 모델 학습에 사용할 데이터 구성
X = digits.images     # 숫자 손글씨 이미지  (1797개)
Y = digits.target     # 실제 숫자(0~9 정답) (10개)

plt.imshow(X[1790], cmap='gray')
plt.show()
print(Y[1790])
```

```python
# * 학습용/테스트용 데이터 분리
X_train, X_test, Y_train, Y_test = train_test_split(
    X.reshape(-1, 64),
    Y,
    test_size = 0.2,      # 전체의 20%를 테스트 데이터 선택
    random_state = 42     # 고정적으로 데이터 선택
)

# print(Y.size)
# print(Y_train.size)
# print(Y_test.size)
```

```python
# * KNN 모델 생성 + 학습
model = KNeighborsClassifier(n_neighbors = 3)
model.fit(X_train, Y_train)
```

```python
# * 예측 진행
pred = model.predict(X_test)

print('예측값:', pred[100])
print('정답:', Y_test[100])

plt.imshow(X_test[100].reshape(8, 8), cmap='gray')
plt.show()
```
